```python


from src.units.wallgenerators.city import CityWallGenerator, EdgeType
import matplotlib.pyplot as plt
```

```python
print("City Wall Generator Demo")

# Create a city wall generator with an initial square 
generator = CityWallGenerator(center_point=(50, 50), size=20, map_size=100)

# Generate the initial square wall
generator.generate_city_wall()
generator.visualize(title="Initial Square")

# Add extensions to create a more complex structure
print("Adding extensions to create a complex wall structure...")
```

```python

# Create a new generator
generator = CityWallGenerator(center_point=(50, 50), size=20, map_size=100)

# Add extensions from different edges
generator.add_square(EdgeType.TOP, source_square_index=0, size=10)  # Add square to top
generator.add_square(EdgeType.RIGHT, source_square_index=0, size=15)  # Add square to right
generator.add_square(EdgeType.BOTTOM, source_square_index=0, size=12)  # Add square to bottom

# Add another square extending from one of our extensions
generator.add_square(EdgeType.RIGHT, source_square_index=1, size=8)  # From the top extension

# Generate the wall outline - note this only shows walls at the perimeter
# Internal shared walls are removed automatically
generator.generate_city_wall()

# Visualize the complex wall structure
generator.visualize(title="Complex City Wall Structure")
```

```python

# Create an even more complex castle-like structure
print("Creating a castle-like structure...")

generator = CityWallGenerator(center_point=(50, 50), size=25, map_size=100)

# Create a symmetrical fortress with corner towers
# Main structure
generator.add_square(EdgeType.TOP, source_square_index=0, size=8)  # North tower
generator.add_square(EdgeType.RIGHT, source_square_index=0, size=8)  # East tower
generator.add_square(EdgeType.BOTTOM, source_square_index=0, size=8)  # South tower 
generator.add_square(EdgeType.LEFT, source_square_index=0, size=8)  # West tower

# Corner towers
generator.add_square(EdgeType.TOP, source_square_index=2, size=5)  # NE corner tower
generator.add_square(EdgeType.TOP, source_square_index=4, size=5)  # NW corner tower
generator.add_square(EdgeType.BOTTOM, source_square_index=2, size=5)  # SE corner tower
generator.add_square(EdgeType.BOTTOM, source_square_index=4, size=5)  # SW corner tower

# Generate and visualize the castle structure
generator.generate_city_wall()
generator.visualize(title="Castle Structure with Corner Towers")

print("Demo completed. Check the generated visualizations.")
```
